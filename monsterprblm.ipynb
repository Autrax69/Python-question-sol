```python


t = int(input())

for _ in range(t):

    n = int(input())
    monsters = list(map(int, input().split()))

    stack = []
    result = []

    for power in monsters:

        # Remove weaker or equal monsters
        while stack and stack[-1] <= power:
            stack.pop()

        # Add current monster
        stack.append(power)

        # Store alive monsters count
        result.append(len(stack))

    print(*result)
```

```python

del input

# Alice Library Problem

s = input("Enter the string: ")

stack = []

for ch in s:
    if ch == '\\':
        temp = []

        while stack and stack[-1] != '/':
            temp.append(stack.pop())

        if stack and stack[-1] == '/':
            stack.pop()

        stack.extend(temp)

    else:
        stack.append(ch)

result = ''.join(c for c in stack if c.isalpha())

print("Final Output:", result)
```

```python
from collections import deque

def solve():
    print("Enter the size of the array (n):")
    n = int(input().strip())

    print(f"Enter the {n} elements of array a (space-separated):")
    array_a = list(map(int, input().strip().split()))

    print(f"Enter the {n} elements of array b (space-separated):")
    array_b = list(map(int, input().strip().split()))

    q_a = deque(array_a)
    total_time = 0


    for target in array_b:

        while q_a[0] != target:
            front_element = q_a.popleft()
            q_a.append(front_element)  # Move front to back
            total_time += 1             # Rotation takes 1 second

        # Front elements now match, remove it from deque a
        q_a.popleft()
        total_time += 1                 # Removal takes 1 second

    print(f"\nMinimum time required: {total_time} seconds")

if __name__ == '__main__':
    solve()
```
