```python


t = int(input())

for _ in range(t):

    n = int(input())
    monsters = list(map(int, input().split()))

    stack = []
    result = []

    for power in monsters:

        # Remove weaker or equal monsters
        while stack and stack[-1] <= power:
            stack.pop()

        # Add current monster
        stack.append(power)

        # Store alive monsters count
        result.append(len(stack))

    print(*result)
```

```python

del input

# Alice Library Problem

s = input("Enter the string: ")

stack = []

for ch in s:
    if ch == '\\':
        temp = []

        while stack and stack[-1] != '/':
            temp.append(stack.pop())

        if stack and stack[-1] == '/':
            stack.pop()

        stack.extend(temp)

    else:
        stack.append(ch)

result = ''.join(c for c in stack if c.isalpha())

print("Final Output:", result)
```
